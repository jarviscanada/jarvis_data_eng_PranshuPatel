### Extracting Fraud Labels JSON file from Storage Account

```python
from pyspark.sql import functions as F

labels_path = "abfss://raw@stfinfraudpranshu01.dfs.core.windows.net/labels/train_fraud_labels.json"

fraud_labels_df = (
    spark.read.text(labels_path)   # read the whole file as raw text
    .select(
        F.from_json(
            F.col("value"),
            "struct<target:map<string,string>>"
        ).alias("json_data")
    )
    .select(
        F.explode(F.col("json_data.target")).alias("transaction_id", "fraud_label")
    )
    .withColumn("is_fraud", F.when(F.col("fraud_label") == "Yes", True).otherwise(False))
)

display(fraud_labels_df.limit(10))
```
